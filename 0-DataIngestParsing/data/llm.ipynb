```python
from llama_index.llms.groq import Groq
from llama_index.core.llms import ChatMessage
```

```python
from dotenv import find_dotenv, load_dotenv
from llama_index.llms.groq import Groq

env_path = find_dotenv(".env", usecwd=True)
if not env_path:
    raise FileNotFoundError("Could not find the project .env file.")
load_dotenv(env_path, override=True)
GROQ_API_KEY = os.getenv("GROQ_API_KEY", "").strip()
if not GROQ_API_KEY:
    raise ValueError("GROQ_API_KEY is missing. Add a new Groq API key to the project .env file.")
if not GROQ_API_KEY.startswith("gsk_"):
    raise ValueError("GROQ_API_KEY does not look like a Groq key. Create a key at https://console.groq.com/keys and replace the value in .env.")

llm = Groq(
    model="openai/gpt-oss-20b",
    api_key=GROQ_API_KEY,
)
```

```python
import textwrap

response = llm.complete("Write a blog post about AI in exactly 50 words.")
print(textwrap.fill(response.text, width=70))
```

```python
message = [ChatMessage(role="system", content="you are an AI assistant"),
           ChatMessage(role="user", content="write a blog post about AI in exactly 50 words.")]

response=llm.stream_chat(message)
for r in response:
    print(r.delta, end="")
```

```python
import os
import textwrap
from dotenv import find_dotenv, load_dotenv
from llama_index.llms.groq import Groq

env_path = find_dotenv(".env", usecwd=True)
if not env_path:
    raise FileNotFoundError("Could not find the project .env file.")
load_dotenv(env_path, override=True)
GROQ_API_KEY = os.getenv("GROQ_API_KEY", "").strip()
if not GROQ_API_KEY:
    raise ValueError("GROQ_API_KEY is missing. Add a Groq API key to the project .env file.")
if not GROQ_API_KEY.startswith("gsk_"):
    raise ValueError("GROQ_API_KEY does not look like a Groq key. Check the GROQ_API_KEY value in .env.")

llm = Groq(
    model="openai/gpt-oss-20b",
    api_key=GROQ_API_KEY,
    temperature=0.2,
    max_output_tokens=200 
)

response = llm.complete("Write a blog post about AI in exactly 50 words.")
print(textwrap.fill(response.text, width=70))
```
